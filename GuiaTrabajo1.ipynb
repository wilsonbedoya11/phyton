## **2.4 Cambiar tipo de datos, limpieza de caracteres**

```python
# Funcion para cambiar quitar puntos
import pandas as pd
import numpy as np

def quitar_puntos_X_coma_dataframe(df):

  for col in df.select_dtypes(include='object'):
    # Cambiar puntos a comas y convertir a float
    try:
      df[col] = df[col].astype(str).str.replace('.', '', regex=False)
    except ValueError:
      # Manejar casos en que la conversión no es posible
      print(f"No se puede convertir la columna {col} a float.")
      # Puedes optar por dejar la columna como está o aplicar otra estrategia de limpieza

  return df

# Aplicar la función de limpieza al DataFrame df_datos
```

```python
# Funcion para cambiar todos los tipos de datos, organziando:
## caracteres especiales se quitan
## cambio de coma por punto
## dejar los campos en porcentajes

import pandas as pd
import numpy as np

def clean_dataframe(df):

  for col in df.select_dtypes(include='object'):
    # Reemplazar símbolos especiales
    df[col] = df[col].astype(str).str.replace(r'[$,€]', '', regex=True)
    #df[col] = df[col].astype(str).str.replace(',', '.', regex=False)

    # Manejar porcentajes
    if df[col].str.contains('%').any():
       df[col] = df[col].str.replace('%', '', regex=False).astype(float)
       df[col] = df[col].astype(str).str.replace(',', '.', regex=False).astype(float) /100

    # Cambiar puntos a comas y convertir a float
    try:
      df[col] = df[col].astype(str).str.replace('.', ',', regex=False).astype(float)
    except ValueError:
      # Manejar casos en que la conversión no es posible
      print(f"No se puede convertir la columna {col} a float.")
      # Puedes optar por dejar la columna como está o aplicar otra estrategia de limpieza

  return df

# Aplicar la función de limpieza al DataFrame df_datos
#df_datos = clean_dataframe(df_datos)
```
